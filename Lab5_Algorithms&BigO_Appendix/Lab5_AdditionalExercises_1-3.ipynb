```python
LAST_NAME = "Cardenas"
STUDENT_ID = "TUPM-26-1258"
FAVORITE_ARTIST = "The 1975"

SEED_DIGIT = int(STUDENT_ID[-1])                                 
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())           
NAME_LENGTH = len(LAST_NAME)                                      
_ARTIST_LETTERS = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(_ARTIST_LETTERS)                            
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in _ARTIST_LETTERS) 


def show_inputs():
    print("=" * 66)
    print("STUDENT-SPECIFIC INPUTS")
    print("=" * 66)
    for name in ("LAST_NAME", "STUDENT_ID", "FAVORITE_ARTIST", "SEED_DIGIT",
                 "ID_SUM", "NAME_LENGTH", "ARTIST_LENGTH", "ARTIST_SUM"):
        print(f"{name:<16}: {globals()[name]}")
    print()


def heading(text):
    print("\n" + "-" * 66 + f"\n{text}\n" + "-" * 66)

if __name__== "__main__":
    show_inputs()
```

```python
# [EXERCISE 1]
def linear_search(a, target):
    comps = 0
    for i, v in enumerate(a):
        comps += 1                   
        if v == target:
            return i, comps
    return -1, comps

def rotated_binary_search(a, target):
    lo, hi, comps = 0, len(a) - 1, 0
    while lo <= hi:
        mid = (lo + hi) // 2
        comps += 1
        if a[mid] == target:
            return mid, comps
        comps += 1
        if a[lo] <= a[mid]:                       
            comps += 2
            if a[lo] <= target and target < a[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:                                   
            comps += 2
            if a[mid] < target and target <= a[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1, comps


def main():
    show_inputs()
    
    n = 20 + NAME_LENGTH
    step = SEED_DIGIT + 3
    original = [1000 + ID_SUM + i * step for i in range(n)]
    k = SEED_DIGIT if SEED_DIGIT > 0 else 1           
    rotated = original[k:] + original[:k]             
    target = original[(ID_SUM + ARTIST_SUM) % n]      
    missing = original[-1] + 1                       

    heading("ORIGINAL DATASET (sorted incident records)")
    print(original)
    heading(f"ROTATED DATASET (first {k} records moved to the end)")
    print(rotated)
    heading("TARGET INCIDENT")
    print(f"Target = {target}   (index {(ID_SUM + ARTIST_SUM) % n} of original; ID_SUM + ARTIST_SUM = {ID_SUM + ARTIST_SUM})")
    print(f"Not-found test target = {missing}")

    heading("SEARCH RESULTS")
    print(f"{'Target':<8}{'Algorithm':<22}{'Position':<10}{'Comparisons'}")
    for t in (target, missing):
        for name, fn in (("Modified Binary", rotated_binary_search), ("Linear Search", linear_search)):
            pos, c = fn(rotated, t)
            print(f"{t:<8}{name:<22}{(pos if pos >= 0 else 'NOT FOUND'):<10}{c}")

    heading("COMPLEXITY ANALYSIS")
    print(f"{'Algorithm':<20}{'Best':<8}{'Average':<10}{'Worst':<10}{'Space'}")
    print(f"{'Modified Binary':<20}{'O(1)':<8}{'O(log n)':<10}{'O(log n)':<10}O(1)")
    print(f"{'Linear Search':<20}{'O(1)':<8}{'O(n)':<10}{'O(n)':<10}O(1)")
    print("Binary search halves the range each step because one half of a rotated"
          "\nsorted list is always sorted; linear search may scan every record.")

    heading("FINAL OUTPUT")
    pos, c = rotated_binary_search(rotated, target)
    _, lc = linear_search(rotated, target)
    print(f"Incident {target} found at position {pos} of the rotated list.")
    print(f"Comparisons -> Modified Binary: {c} | Linear: {lc} (n = {n})")


if __name__ == "__main__":
    main()
```

```python
# [EXERCISE 2]
from collections import Counter

def merge_sorted(a, b):
    i = j = comps = 0
    out = []
    while i < len(a) and j < len(b):
        comps += 1                        
        if a[i] <= b[j]:
            out.append(a[i]); i += 1
        else:
            out.append(b[j]); j += 1
    out.extend(a[i:]); out.extend(b[j:])   
    return out, comps


def main():
    show_inputs()
    n_a = 8 + SEED_DIGIT
    n_b = 4 + NAME_LENGTH
    A = [2000 + ID_SUM + i * (SEED_DIGIT + 3) for i in range(n_a)]
    B = [2000 + ARTIST_LENGTH + i * (NAME_LENGTH // 2 + 2) for i in range(n_b)]

    heading(f"DEPARTMENT A DATASET (SEED_DIGIT={SEED_DIGIT}, ID_SUM={ID_SUM}) - {len(A)} records")
    print(A)
    heading(f"DEPARTMENT B DATASET (ARTIST_LENGTH={ARTIST_LENGTH}, NAME_LENGTH={NAME_LENGTH}) - {len(B)} records")
    print(B)

    merged, comps = merge_sorted(A, B)
    heading("CONSOLIDATED DATASET")
    print(merged)

    heading("NUMBER OF RECORDS BEFORE AND AFTER CONSOLIDATION")
    print(f"Before: A = {len(A)}, B = {len(B)}, total = {len(A) + len(B)}")
    print(f"After : {len(merged)}")

    heading("COMPARISON COUNT")
    print(f"Comparisons performed: {comps}  (bounds: min {min(len(A), len(B))}, max {len(A) + len(B) - 1})")

    is_sorted = all(merged[i] <= merged[i + 1] for i in range(len(merged) - 1))
    same_count = len(merged) == len(A) + len(B)
    same_items = Counter(merged) == Counter(A + B)
    heading("ORDERING VERIFICATION")
    print(f"Ascending order : {'PASSED' if is_sorted else 'FAILED'}")
    print(f"Record count    : {'PASSED' if same_count else 'FAILED'}")
    print(f"No records lost : {'PASSED' if same_items else 'FAILED'}")
    dups = sorted(v for v, c in Counter(merged).items() if c > 1)
    print(f"Duplicate IDs across departments (kept): {dups if dups else 'none'}")

    heading("TIME AND SPACE COMPLEXITY")
    print("Time : O(n + m) - each record is examined once (n = |A|, m = |B|)")
    print("Space: O(n + m) - a new consolidated list is built")

    heading("FINAL OUTPUT")
    print(f"{len(merged)} enrollment records consolidated in order using {comps} comparisons.")


if __name__ == "__main__":
    main()
```

```python
# [EXERCISE 3]
import time

REPS = 300  

def bubble_sort(arr):
    a, comps, moves = list(arr), 0, 0
    for i in range(len(a) - 1):
        swapped = False
        for j in range(len(a) - 1 - i):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]; moves += 2; swapped = True
        if not swapped:
            break
    return a, comps, moves

def selection_sort(arr):
    a, comps, moves = list(arr), 0, 0
    for i in range(len(a) - 1):
        m = i
        for j in range(i + 1, len(a)):
            comps += 1
            if a[j] < a[m]:
                m = j
        if m != i:
            a[i], a[m] = a[m], a[i]; moves += 2
    return a, comps, moves

def insertion_sort(arr):
    a, comps, moves = list(arr), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comps += 1
            if a[j] > key:
                a[j + 1] = a[j]; moves += 1; j -= 1
            else:
                break
        if j + 1 != i:
            a[j + 1] = key; moves += 1
    return a, comps, moves

def avg_time_us(fn, data):
    start = time.perf_counter()
    for _ in range(REPS):
        fn(data)
    return (time.perf_counter() - start) / REPS * 1e6

def main():
    show_inputs()
    n, m = 20 + NAME_LENGTH, 101               
    mult = ARTIST_SUM % 100 + 1
    random_orders = [1000 + (i * mult + ID_SUM * SEED_DIGIT + ID_SUM) % m for i in range(n)]
    assert len(set(random_orders)) == n
    inputs = {
        "Sorted": sorted(random_orders),
        "Reverse": sorted(random_orders, reverse=True),
        "Random": random_orders,
    }

    heading("GENERATED ORDER DATASET (n = %d)" % n)
    print(random_orders)
    heading("SORTED, REVERSE, AND RANDOM INPUTS (same orders)")
    for k, v in inputs.items():
        print(f"{k:<8}: {v}")

    algos = {"Bubble": bubble_sort, "Selection": selection_sort, "Insertion": insertion_sort}
    expected = sorted(random_orders)

    results = []
    for cond, data in inputs.items():
        for name, fn in algos.items():
            out, comps, moves = fn(data)
            t = avg_time_us(fn, data)
            results.append((cond, name, comps, moves, t, out == expected))

    heading("COMPARISON COUNTS, MOVEMENT COUNTS, EXECUTION TIMES, VERIFICATION")
    print(f"{'Input':<9}{'Algorithm':<11}{'Comparisons':>12}{'Movements':>11}{'Time (us)':>12}  Verified")
    for cond, name, c, mv, t, ok in results:
        print(f"{cond:<9}{name:<11}{c:>12}{mv:>11}{t:>12.2f}  {'PASSED' if ok else 'FAILED'}")
    print("\nMovement = element writes (a swap = 2 writes; an insertion shift/placement = 1).")
    print(f"Time = average of {REPS} runs on this machine.")

    heading("PERFORMANCE ACROSS INPUT CONDITIONS (fewest comparisons)")
    for cond in inputs:
        best = min((r for r in results if r[0] == cond), key=lambda r: r[2])
        print(f"{cond:<9}-> {best[1]} ({best[2]} comparisons)")

    heading("BIG O AND CASE ANALYSIS")
    print(f"{'Algorithm':<11}{'Best':<9}{'Average':<9}{'Worst':<9}{'Space':<7}Best-case input")
    print(f"{'Bubble':<11}{'O(n)':<9}{'O(n^2)':<9}{'O(n^2)':<9}{'O(1)':<7}Sorted (early exit after 1 pass)")
    print(f"{'Selection':<11}{'O(n^2)':<9}{'O(n^2)':<9}{'O(n^2)':<9}{'O(1)':<7}None - always scans the unsorted part")
    print(f"{'Insertion':<11}{'O(n)':<9}{'O(n^2)':<9}{'O(n^2)':<9}{'O(1)':<7}Sorted (one comparison per element)")
    print(f"\nFor n = {n}: n(n-1)/2 = {n * (n - 1) // 2} comparisons is the quadratic worst-case reference.")
    print("Observations:")
    print(" - Sorted (best case): Bubble and Insertion need only n-1 comparisons; Selection still needs n(n-1)/2.")
    print(" - Reverse (worst case): all three approach n(n-1)/2 comparisons; Bubble and Insertion also move the most.")
    print(" - Random (average case): quadratic growth for all; Selection uses the fewest movements (at most n-1 swaps).")

    heading("FINAL OUTPUT")
    print(f"All {len(results)} sorting runs verified: {all(r[5] for r in results)}")


if __name__ == "__main__":
    main()
```
